# Install a GitHub package with uv

This experiment tests whether uv can install a tiny Python package from a fixed Git commit into the running Colab kernel. No separate virtual environment is created. Run all cells in order; rerunning them is safe. The package has no runtime dependencies.

The publisher must replace `YOUR_GITHUB_USERNAME` before opening this notebook in Colab. The commit pins the package source; Python, uv, and build tools are not locked.


```python
import importlib.util
import subprocess
import sys

if importlib.util.find_spec("uv") is None:
    subprocess.check_call([sys.executable, "-m", "pip", "install", "uv"])
```

```python
PACKAGE_URL = "git+https://github.com/suhelhammoud/collab-test.git@9265cdd0dbb1d2af544685454b40bc1caedd74ff"

subprocess.check_call([
    sys.executable, "-m", "uv", "pip", "install",
    "--python", sys.executable, PACKAGE_URL,
])
```
